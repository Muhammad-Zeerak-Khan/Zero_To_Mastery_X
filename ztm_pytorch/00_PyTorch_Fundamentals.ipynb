## Introduction to Tensors

### Creating Tensors

```python
# Let's define a helper function to get the dimension and shape of the input tensor
def get_tensor_info(input_tensor):
    ndim = input_tensor.ndim
    shape = input_tensor.shape
    dtype = input_tensor.dtype
    print(f"The dim of the input tensor is : {ndim}")
    print(f"The shape of the input tensor is : {shape}")
    print(f"The data type of the input tensor is : {dtype}")

    if ndim == 0:
        print(f"The value of the input tensor is : {input_tensor.item()}")
    else:
        print(f"The value of the input tensor is : {input_tensor}")
```

```python
# Scaler
scaler = torch.tensor(5)
get_tensor_info(scaler)
```

```python
# Let's create a vector
vector = torch.tensor([1, 2])
get_tensor_info(vector)
```

## Common Pitfalls
**NOTE** Tensor datatypes is one of the 3 big errors to run into with PyTorch and DL
1. Tensors not having the right datatypes. Two tensors having different datatypes.
2. Tensors not having the right shape.
3. Tensors not having the same device. e.g. one on CPU and other on GPU.

```python
float_16_tensor = torch.rand(size=(2, 4), dtype=torch.float16)
float_32_tensor = torch.rand(size=(2, 4), dtype=torch.float32)
get_tensor_info(float_16_tensor)
get_tensor_info(float_32_tensor)
```

```python
# Let's try some mathematical operations on  two tensors of different dtypes
addition = float_16_tensor + float_32_tensor
get_tensor_info(addition)
subtraction = float_16_tensor - float_32_tensor
get_tensor_info(subtraction)
mutliplication = float_16_tensor * float_32_tensor
get_tensor_info(mutliplication)
division = float_16_tensor / float_32_tensor
get_tensor_info(division)
```

```python
# Helper function the get the position of the min and max value in the given tensor


from torch._tensor import Tensor


def get_argmin_and_argmax_value(
    given_tensor: torch.Tensor,
) -> tuple[Tensor, Tensor]:
    """Returns the position of the max and min value in the given tensor"""
    print(f"The given tensor : {given_tensor}")
    print(f"The max value is at position : {torch.argmax(given_tensor)}")
    max_value_position: torch.Tensor = torch.argmax(given_tensor)
    print(f"The min value is at position : {torch.argmin(given_tensor)}")
    min_value_position: torch.Tensor = torch.argmin(given_tensor)
    return max_value_position, min_value_position
```

```python
max_pos, min_pos = get_argmin_and_argmax_value(x)
max_pos, min_pos
```

## Tensor Shape Manipulation
* Reshaping: Reshapes the input tensor to the desired shape.
* View:  Returns a view of the input tensor of a certain shape but keep the same memory as the original tensor.
* Stacking: Combine multiple tensors on top of each other either horizontally (hstack) or vertically (vstack).
* Squeeze: Remove all `1` dimensions from a tensor.
* Unsqueeze: Add a `1` dimension to a tensor.
* Permute: Reterns a view of the input with dimensions permuted (swapped) in a certain way.

```python
# Let's create a tensor
x = torch.arange(1, 11)
x, x.shape
```

```python
# Reshape
x_reshaped = x.reshape(10, 1)
x_reshaped, x_reshaped.shape
```

```python
# View
# Remember that view shares the same memoery as the original
# Meaning: A change in 'z' would be reflected to x as well

z = x.view(2, 5)
z, z.shape
```

```python
z[1, 2] = 14
print(f"The updated value of z is : {z}")
print(f"The value of x also got updated  : {x}")
```

```python
# Stack tensors on top
x_horizontally_stacked = torch.stack([x, x, x, x], dim=0)
x_horizontally_stacked
```

```python
x_vertically_stacked = torch.stack([x, x, x, x], dim=1)
x_vertically_stacked
```

```python
# Squeeze - Removes all the `1` dimensions from the input tensor
x_reshaped, x_reshaped.shape
```

```python
a = torch.rand(2, 1, 3, 4, 1, 6)
# a,
a.shape
```

```python
a_squeezed = a.squeeze()
# a_squeezed
a_squeezed.shape
```

The argument `dim` suppors a list of dimensions on which the `1` dimension would be squeezed.
NOTE: 
* If the dim passed is not zero, nothing happens.
* If no dim is passed, all the dimension with `1` would be removed (squeezed).
* If a valid argument for the squeeze is provided, then only that `1` dimension would be squeezed, other `1` dimensions would remain the same. 
* Squeeze operation shares the same memory as the original and therefore alters the original tensor as well, so beware of it.

```python
# Let's unsqueeze
print(f"The shape of the squeezed tensor: {a_squeezed.shape}")

a_unsqueezed = a_squeezed.unsqueeze(dim=2)
print(f"The shape of the unsqueezed tensor : {a_unsqueezed.shape}")
```

```python
# Let's permute the tensor
b = torch.randn(size=(3, 6, 7, 4, 5, 2))
# b,
b.shape
```

```python
b_permuted = b.permute(4, 2, 0, 5, 3, 1)
print(f"The shape of the tensor before permuation : {b.shape}")
print(f"The shape of the tensor after permuation : {b_permuted.shape}")
```

### Common Application of the permute function is in images, where (H,W,C) has to be permuted to (C,H,W)

```python
original_image_tensor: torch.Tensor = torch.rand(size=(224, 224, 3))
print(
    f"The shape of the original image tensor : {original_image_tensor.shape} (H x W x C)"
)
permuted_image_tensor: torch.Tensor = original_image_tensor.permute(dims=(2, 0, 1))
print(
    f"The shape of the permuted image tensor : {permuted_image_tensor.shape} (C x H x W)"
)
```

```python
permuted_image_tensor[1][220][10] = 0.2525
permuted_image_tensor[1][220][10]
```

```python
original_image_tensor[:, :, 1][220][10] = 0.349
original_image_tensor[:, :, 1][220][10]
```

```python
original_image_tensor[:, :, 1][220][10]
```

```python
permuted_image_tensor: torch.Tensor = original_image_tensor.permute(dims=(2, 0, 1))
```

```python
permuted_image_tensor[1][220][10]
```

NOTE:
* As permute returns a view of the original tensor i.e. it shares the same memory location, a change in the permuted tensor results 
  a change in the original tensor as well. 

```python
# Numpy to PyTorch and vice versa
np_array = np.arange(1.0, 10.0)
print(
    f"Numpy array has the following shape and dtype respectively : {np_array.shape}, {np_array.dtype}"
)
torch_array = torch.from_numpy(np_array)
print(
    f"The torch array converted from numpy has the following shape and dtype respectively : {torch_array.shape}, {torch_array.dtype}"
)
print("This means that torch takes the data type from the numpy array")
```

```python
# To change this behavior
torch_array_float_32 = torch.from_numpy(np_array).type(torch.float32)
print(
    f"The explicitly created torch tensor now has the dtype : {torch_array_float_32.dtype}"
)
```

```python
## Reproducibility
tensor_a = torch.rand(4, 5)
tensor_b = torch.rand(4, 5)

# Lets compare the two
print(f"Tensor A: {tensor_a}")
print(f"Tensor B: {tensor_b}")
print(f"Are the two same ? \n {tensor_a == tensor_b}")
```

```python
RANDOM_SEED = 88
torch.manual_seed(RANDOM_SEED)
tensor_c = torch.rand(4, 5)
tensor_d = torch.rand(4, 5)


# Lets compare the two
print(f"Tensor C: {tensor_c}")
print(f"Tensor D: {tensor_d}")
print(f"Are the two same ? \n {tensor_c == tensor_d}")

print(
    "Turns out the seed has to be set right before the operation for each such operation. Let's give it another shot, shall we?"
)
```

```python
RANDOM_SEED = 88
torch.manual_seed(RANDOM_SEED)
tensor_c = torch.rand(4, 5)
torch.manual_seed(RANDOM_SEED)
tensor_d = torch.rand(4, 5)

# Lets compare the two
print(f"Tensor C: {tensor_c}")
print(f"Tensor D: {tensor_d}")
print(f"Are the two same ? \n {tensor_c == tensor_d}")

print(
    "Wolla! The two tensor created are the same now. \n This solves the concern of reproducibility."
)
```
