### 2. Create a random tensor with shape `(7, 7)`.


````python
# Import torch
import torch


# Create random tensor
random_tensor = torch.rand(size=(7, 7))
random_tensor, random_tensor.shape
````

### 3. Perform a matrix multiplication on the tensor from 2 with another random tensor with shape `(1, 7)` (hint: you may have to transpose the second tensor).

````python
# Create another random tensor
second_random_tensor = torch.rand(size=(1, 7))
print(f"Second random tensor's shape before transpose: {second_random_tensor.shape}")
# Transpose the second random tensor inorder to perform matrix multiplication
second_random_tensor_transposed = second_random_tensor.T
print(f"Second random tensor's shape after transpose: {second_random_tensor.shape}")

# let's print the shape of the two tensors to be multiplied
print(f"Shape of the first tensor : {random_tensor.shape}")
print(f"Shape of the second tensor : {second_random_tensor.shape}")
print("Matrix mutiplication now possible")

# Perform matrix multiplication
product = torch.matmul(random_tensor, second_random_tensor_transposed)
print(f"Product of the two tensors : \n{product}")
print(f"Product of the two tensors has the shape : {product.shape}")
````


### 10. Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the `1` dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

The output should look like:

```
tensor([[[[0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297,
           0.3653, 0.8513]]]]) torch.Size([1, 1, 1, 10])
tensor([0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297, 0.3653,
        0.8513]) torch.Size([10])
```

````python
# Set seed
torch.manual_seed(7)


# Create random tensor
random_tensor = torch.rand(1, 1, 1, 10)
print(random_tensor, random_tensor.shape)


# Remove single dimensions
random_tensor_squeezed = random_tensor.squeeze()

# Prinprint(f"Random tensor shape: {random_tensor.shape}")
print(f"random_tensor_squeezed: {random_tensor_squeezed}")
print(f"Random tensor shape after squeeze: {random_tensor_squeezed.shape}")
````
